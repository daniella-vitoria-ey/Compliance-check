```python
from src.rag.retrieval import retrieve
from src.rag.reranker import rerank
```

```python
queries = [
    "investimento agressivo sem risco",
    "cliente conservador investindo em ações",
    "produto com garantia de retorno"
]

for q in queries:
    print("\n==============================")
    print("QUERY:", q)

    docs = retrieve(q)

    print("\nANTES (retrieve):")
    for d in docs:
        print("-", d["text"])

    docs_reranked = rerank(q, docs)

    print("\nDEPOIS (rerank):")
    for d in docs_reranked:
        print("-", d["text"])
```

# Análise dos Resultados do Re-ranking

## Observação 1 – Query: "investimento agressivo sem risco"

O resultado do re-ranking manteve praticamente a mesma ordem dos documentos retornados pelo retrieval inicial. Isso indica que, para esta consulta, a recuperação inicial já estava bem alinhada com a intenção da query, não exigindo grandes ajustes na ordenação.

---

## Observação 2 – Query: "cliente conservador investindo em ações"

Para esta consulta, o re-ranking trouxe melhorias perceptíveis. Os documentos mais relevantes passaram a incluir regras explícitas de comunicação e restrições relacionadas a risco, o que é mais adequado para a análise de conformidade. Isso mostra que o re-ranking ajudou a priorizar conteúdos mais alinhados à tarefa.

---

## Conclusão

A etapa de re-ranking se mostrou útil para refinar a ordenação dos documentos em alguns casos, especialmente quando a recuperação inicial traz resultados menos organizados ou parcialmente relevantes.

No entanto, em consultas onde o retrieval já retorna documentos bem alinhados, o impacto do re-ranking é limitado.

A abordagem atual utiliza TF-IDF, que é eficiente para prototipagem, mas pode ser aprimorada com embeddings semânticos mais avançados em futuras evoluções.
